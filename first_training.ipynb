```python
import pandas as pd 
from logistic_regression import gradient_descent, predict
from metrics import confusion_matrix, false_positive_rate
```

```python
from sklearn.model_selection import train_test_split

def calculate_logistic_regression(df):

    X_df = df.drop(columns=["target"])
    y_df = (df["target"] == 1).astype(int)

    X_train, X_test, y_train, y_test = train_test_split(
        X_df.values.astype(float),
        y_df.values,
        test_size=0.2,
        stratify=y_df.values,
        shuffle=True,
        random_state=42,
    )

    weight, bias = gradient_descent(X_train, y_train)

    y_pred = predict(X_test, weight, bias)
    matrix = confusion_matrix(y_test, y_pred)
    print("confusion matrix [[TN, FP], [FN, TP]]")
    print(matrix)
    print(f"FPR: {false_positive_rate(matrix):.4f}")
```

```python
for i in range(1, 21):
    try:
        df = pd.read_csv(f"class_imbalance_refactored/df{i}.csv")
        print(f"df{i}")
        calculate_logistic_regression(df)
    except Exception as e:
        continue
```

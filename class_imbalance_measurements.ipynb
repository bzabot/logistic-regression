Problema a resolver:

Num problema de classificação binária com alto desbalanceamento de classes, os algoritmos de classificação tender ter um alto erro/bias para a classe majoritária
Ex:
Se em 1000 exemplos, 950 são positivos, e 50 negativos. Os modelos tendem a prever mais o positivo.

1) Como medir isso?
No caso do exemplo acima, o número de erros pelo class imbalance é igual ao número de falsos positivos. I.e., previu positivo mas na vdd é negativo.


```python
df = pd.read_csv("./class_imbalance/dataset_1039_hiva_agnostic.csv")
df
```

De 4229 linhas, apenas 149 linhas tem label = 1

Se nós tratarmos: -1 como positivo e 1 como negativo, podemos treinar o modelo. Para esse dataset nao é necessário tratar os dados

```python
split_idx = int(len(df) * 0.8)

X_df = df.drop(columns=["label"])
y_df = (df["label"] == 1).astype(int)

X_train = X_df.iloc[:split_idx].values.astype(float)
X_test = X_df.iloc[split_idx:].values.astype(float)

y_train = y_df.iloc[:split_idx].values
y_test = y_df.iloc[split_idx:].values
```

```python
from logistic_regression import gradient_descent

weight, bias = gradient_descent(X_train, y_train)

print(weight, bias)
```

```python
from logistic_regression import predict

y_pred = predict(X_test, weight, bias)
len(y_pred)
```

```python
from metrics import confusion_matrix, false_positive_rate

matrix = confusion_matrix(y_test, y_pred)
print("confusion matrix [[TN, FP], [FN, TP]]")
print(matrix)
print(f"FPR: {false_positive_rate(matrix):.4f}")
```

That is, we got 35 FNs in 846 predictions. That means, 35 values should be positive but were predicted as negative.

```python
## Rename -> target
df = df.rename(columns={"c": "target"})
df
```

```python
import numpy as np
import pandas as pd

# Remove accidental index columns
df = df.loc[:, ~df.columns.str.contains("^Unnamed")]

# Normalize column names
df.columns = df.columns.str.strip()

# Drop constant columns
constant_columns = [
    col for col in df.columns
    if col != "target" and df[col].nunique(dropna=False) <= 1
]

df = df.drop(columns=constant_columns)

# Split temporarily
y_df = df["target"]
X_df = df.drop(columns=["target"])

# Fill categorical missing values before dummies
categorical_columns = X_df.select_dtypes(include=["object", "string"]).columns

for col in categorical_columns:
    X_df[col] = X_df[col].astype("string").str.strip()
    X_df[col] = X_df[col].fillna("__missing__")

# Generate dummy variables
X_df = pd.get_dummies(
    X_df,
    columns=categorical_columns,
    drop_first=True,
    dtype=float,
)

# Fill numeric missing values
X_df = X_df.fillna(X_df.mean(numeric_only=True))
X_df = X_df.fillna(0)

# Drop constant dummy/numeric columns after encoding
X_df = X_df.loc[:, X_df.nunique(dropna=False) > 1]

# Put target back as the final column
df_clean = pd.concat(
    [X_df.reset_index(drop=True), y_df.reset_index(drop=True).rename("target")],
    axis=1,
)
```

```python
## Majority class -> 1 and Minority class -> 0
df_clean["target"] = (df_clean["target"] == False).astype(int)

df_clean
```

```python
df_clean.to_csv("class_imbalance_refactored/df20.csv", index=False)
```

```python
def regenerate_ipums_dataset(source_file, output_file):
    source = pd.read_csv(f"class_imbalance/{source_file}")
    source = source.loc[:, ~source.columns.str.contains("^Unnamed")]
    source.columns = source.columns.str.strip()

    y = source["binaryClass"].map({"N": 0, "P": 1})
    if y.isna().any():
        raise ValueError(f"Unexpected labels in {source_file}: {source['binaryClass'].unique()}")

    X = source.drop(columns=["binaryClass"])
    categorical_columns = X.select_dtypes(include=["object", "string"]).columns

    for col in categorical_columns:
        X[col] = X[col].astype("string").str.strip().fillna("__missing__")

    X = pd.get_dummies(X, columns=categorical_columns, drop_first=True, dtype=float)
    X = X.apply(pd.to_numeric, errors="coerce")
    X = X.fillna(X.mean(numeric_only=True)).fillna(0)
    X = X.loc[:, X.nunique(dropna=False) > 1]

    df_clean = pd.concat(
        [X.reset_index(drop=True), y.reset_index(drop=True).rename("target")],
        axis=1,
    )
    df_clean.to_csv(f"class_imbalance_refactored/{output_file}", index=False)
    return df_clean["target"].value_counts()

print("df3", regenerate_ipums_dataset("dataset_1018_ipums_la_99-small.csv", "df3.csv"))
print("df4", regenerate_ipums_dataset("dataset_1002_ipums_la_98-small.csv", "df4.csv"))
```

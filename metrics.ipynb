```python
import pandas as pd
```

### Fit the model into the logistic regression

```python
df = pd.read_csv('./class_imbalance_refactored/df3.csv')
df
```

```python
train = df.sample(frac=0.8, random_state=42)
test = df.drop(train.index)
```

```python
X_train = train.drop(columns=["target"]).values.astype(float)
y_train = train["target"].values.astype(float)
```

```python
from logistic_regression import gradient_descent

weight, bias = gradient_descent(X_train, y_train)

print(weight, bias)
```

```python
from logistic_regression import predict
X_test = test.drop(columns=["target"]).values.astype(float)
y_test = test["target"].values.astype(float)
y_pred = predict(X_test, weight, bias)
```

```python
from metrics import confusion_matrix

matrix = confusion_matrix(y_test, y_pred)
matrix
```

### 1. Accuracy Gain Over Majority Baseline
Qual o ganho do modelo quando comparado a uma modelo que prevê apenas a maioria.

```python
from metrics import accuracy

# Numero total da maioria / 
acc_majority = matrix[1][1] / len(y_test)
acc = accuracy(matrix)

acc_gain = acc - acc_majority
acc_gain
```

### 2. Balanced Accuracy
BA = 0,5 para o classificador trivial, qualquer BA > 0.5 indica aprendizado real


```python
from metrics import get_counts
tp, fp, fn, tn = get_counts(matrix)

balanced_accuracy = (tp/(tp+fn) + tn/(tn+fp))/2
balanced_accuracy
```

### 3. Matthews Correlation Coefficient

```python
from numpy import sqrt


tp, fp, fn, tn = get_counts(matrix)
mcc = (tp*tn - fp*fn) / sqrt((tp+fp)*(tp+fn)*(tn+fp)*(tn+fn))
mcc
```

tn e fn é 0 para o nosso caso, fazendo com que o denominador seja 0 -> nan
